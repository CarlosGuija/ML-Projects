# Sentiment Classifier Exploration

This notebook is used for exploratory data analysis (EDA) on the sentiment classification dataset. The goal is to understand the data better and derive insights that can inform the model development process.

```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set(style='whitegrid')
```

```python
# Load the dataset
data = pd.read_csv('../data/raw/dataset.csv')

# Display the first few rows of the dataset
data.head()
```

```python
# Summary of the dataset
data.info()

# Check for missing values
data.isnull().sum()
```

```python
# Visualize the distribution of sentiment labels
plt.figure(figsize=(8, 6))
sns.countplot(x='label', data=data)
plt.title('Distribution of Sentiment Labels')
plt.xlabel('Sentiment Label')
plt.ylabel('Count')
plt.show()
```

```python
# Analyze the length of text data
data['text_length'] = data['text'].apply(len)

plt.figure(figsize=(10, 6))
sns.histplot(data['text_length'], bins=30, kde=True)
plt.title('Text Length Distribution')
plt.xlabel('Length of Text')
plt.ylabel('Frequency')
plt.show()
```

## Insights

1. The dataset contains a balanced distribution of sentiment labels.
2. The text length varies significantly, which may impact model performance.

Further analysis can be conducted to explore the relationship between text length and sentiment.
